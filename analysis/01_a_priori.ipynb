### Decisões a priori e definições

| Decisão | Valor |
|---------|-------|
| Unidade de análise | 19 decisões binárias por avaliação (avaliação × padrão) |
| Coeficientes de fiabilidade | α de Krippendorff, κ por pares, AC1 de Gwet, P_pos/P_neg, por padrão |
| Limiares de aceitação | α ≥ 0,667 a título provisório; ≥ 0,80 satisfatório (o mínimo de concordância para confiar num padrão) |
| Métricas de validade | precisão, revocação, F1 por (modelo, padrão), em média macro |
| Incerteza | intervalos de confiança a 95% por *bootstrap* |
| Referência | consenso adjudicado (as decisões finais do autor), com a influência do painel medida por uma passagem cega |

```python
LIMIARES = {"alpha_provisorio": 0.667, "alpha_satisfatorio": 0.80,
            "kappa_moderado": (0.41, 0.60), "bootstrap_n": 2000, "ic": 0.95}
for k, v in LIMIARES.items():
    print(f"{k:20s} {v}")
```

## Definições

| | Definição |
|-|-----------|
| Fiabilidade | Pergunta se os modelos do painel concordam entre si (α, κ, AC1). A adjudicação não entra neste cálculo |
| Validade | Pergunta se cada modelo acerta face ao autor (precisão, revocação, F1 contra a adjudicação) |
| Direcção da adjudicação | Pergunta em que sentido vão as divergências quando o painel e o autor discordam |
